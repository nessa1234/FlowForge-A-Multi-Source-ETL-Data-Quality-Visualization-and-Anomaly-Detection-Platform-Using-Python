```python

# ====================================================
# FLOWFORGE - STEP 4.1
# MISSING VALUE ANALYSIS
# ====================================================

def missing_value_report(data):

    missing_count = data.isnull().sum()

    missing_percentage = (
        missing_count / len(data) * 100
    ).round(2)

    report = pd.DataFrame({
        "Missing Values": missing_count,
        "Missing Percentage (%)": missing_percentage
    })

    report = report.sort_values(
        by="Missing Values",
        ascending=False
    )

    return report


missing_report = missing_value_report(df)

print("FLOWFORGE - MISSING VALUE REPORT")
display(missing_report)
```

```python

# ====================================================
# FLOWFORGE - STEP 4.4
# INVALID VALUES & OUTLIER INVESTIGATION
# ====================================================

import pandas as pd

negative_quantity = (df["Quantity"] < 0).sum()
zero_quantity = (df["Quantity"] == 0).sum()

negative_price = (df["UnitPrice"] < 0).sum()
zero_price = (df["UnitPrice"] == 0).sum()

# Invoice numbers starting with C indicate cancellations
cancelled_invoices = (
    df["InvoiceNo"].astype(str).str.startswith("C")
).sum()

quality_checks = pd.DataFrame({
    "Quality Check": [
        "Negative Quantity",
        "Zero Quantity",
        "Negative UnitPrice",
        "Zero UnitPrice",
        "Cancelled Invoice Rows"
    ],
    "Affected Records": [
        negative_quantity,
        zero_quantity,
        negative_price,
        zero_price,
        cancelled_invoices
    ]
})

quality_checks["Percentage (%)"] = (
    quality_checks["Affected Records"]
    / len(df) * 100
).round(2)

print("FLOWFORGE - SUSPICIOUS VALUE REPORT")
display(quality_checks)
```

```python

# ====================================================
# FLOWFORGE - STEP 4.5
# TRANSACTION CLASSIFICATION
# ====================================================

import pandas as pd
import numpy as np

# Create working copy - preserve original data
df_processed = df.copy(deep=True)

# Identify cancelled invoices
df_processed["Is_Cancelled"] = (
    df_processed["InvoiceNo"]
    .astype(str)
    .str.upper()
    .str.startswith("C")
)

# Classify transaction records
conditions = [
    df_processed["UnitPrice"] < 0,

    (df_processed["Quantity"] < 0)
    & (~df_processed["Is_Cancelled"]),

    df_processed["Is_Cancelled"],

    df_processed["Quantity"] == 0,

    df_processed["UnitPrice"] == 0
]

categories = [
    "Negative Price - Review",
    "Negative Quantity - Review",
    "Cancelled Transaction",
    "Zero Quantity - Review",
    "Zero Price - Review"
]

df_processed["Transaction_Status"] = np.select(
    conditions,
    categories,
    default="Valid Sale"
)

# Summary report
status_report = (
    df_processed["Transaction_Status"]
    .value_counts()
    .reset_index()
)

status_report.columns = [
    "Transaction Status",
    "Record Count"
]

status_report["Percentage (%)"] = (
    status_report["Record Count"]
    / len(df_processed) * 100
).round(2)

print("FLOWFORGE - TRANSACTION CLASSIFICATION")
display(status_report)
```

```python

# ====================================================
# FLOWFORGE - TRANSACTION STATUS VISUALIZATION
# ====================================================

import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(11, 5))

sns.barplot(
    data=status_report,
    x="Transaction Status",
    y="Record Count",
    color="steelblue"
)

plt.title("FlowForge - Transaction Classification")
plt.xlabel("Transaction Status")
plt.ylabel("Number of Records")

plt.xticks(rotation=30, ha="right")

plt.tight_layout()
plt.show()
```

```python

# ====================================================
# FLOWFORGE - STEP 4.7
# OUTLIER DETECTION USING IQR
# ====================================================

import pandas as pd

# Use positive-price, positive-quantity sales
# to establish meaningful IQR boundaries.
sales_df = df_processed[
    (df_processed["Transaction_Status"] == "Valid Sale")
].copy()

def detect_outliers_iqr(data, column):
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outlier_mask = (
        (data[column] < lower_bound) |
        (data[column] > upper_bound)
    )

    return {
        "Column": column,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "Lower Bound": lower_bound,
        "Upper Bound": upper_bound,
        "Outlier Count": int(outlier_mask.sum()),
        "Outlier Percentage (%)": round(
            outlier_mask.mean() * 100, 2
        )
    }

outlier_report = pd.DataFrame([
    detect_outliers_iqr(sales_df, "Quantity"),
    detect_outliers_iqr(sales_df, "UnitPrice")
])

print("FLOWFORGE - IQR OUTLIER REPORT")
display(outlier_report)
```

```python

# ====================================================
# FLOWFORGE - OUTLIER VISUALIZATION
# ====================================================

import matplotlib.pyplot as plt
import seaborn as sns

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.boxplot(
    x=sales_df["Quantity"],
    ax=axes[0],
    color="skyblue"
)

axes[0].set_title("Quantity Outlier Analysis")

sns.boxplot(
    x=sales_df["UnitPrice"],
    ax=axes[1],
    color="lightgreen"
)

axes[1].set_title("Unit Price Outlier Analysis")

plt.suptitle("FlowForge - IQR Outlier Detection")
plt.tight_layout()
plt.show()
```

```python

# ====================================================
# FLOWFORGE - STEP 5.1
# DATA TYPE CONVERSION
# ====================================================

import pandas as pd

# Preserve our transaction classifications
df_cleaned = df_processed.copy(deep=True)

# Convert invoice date
df_cleaned["InvoiceDate"] = pd.to_datetime(
    df_cleaned["InvoiceDate"],
    format="%m/%d/%Y %H:%M",
    errors="coerce"
)

# Convert CustomerID to nullable string
df_cleaned["CustomerID"] = (
    pd.to_numeric(
        df_cleaned["CustomerID"],
        errors="coerce"
    )
    .astype("Int64")
    .astype("string")
)

# Ensure identifiers remain strings
df_cleaned["InvoiceNo"] = (
    df_cleaned["InvoiceNo"].astype("string")
)

df_cleaned["StockCode"] = (
    df_cleaned["StockCode"].astype("string")
)

print("FLOWFORGE - DATA TYPE CONVERSION COMPLETED")

print("\nUpdated Data Types:")
print(df_cleaned.dtypes)

print("\nInvalid/Missing Invoice Dates:")
print(df_cleaned["InvoiceDate"].isna().sum())

print("\nDataset Preview:")
display(df_cleaned.head())
```

```python

# ====================================================
# FLOWFORGE - STEP 5.2
# MISSING VALUE TREATMENT
# ====================================================

import pandas as pd

# Work with the cleaned dataset from Step 5.1
df_cleaned = df_cleaned.copy(deep=True)

# Flag missing CustomerIDs
df_cleaned["Missing_CustomerID"] = (
    df_cleaned["CustomerID"].isna()
)

# Flag missing or blank descriptions
df_cleaned["Missing_Description"] = (
    df_cleaned["Description"].isna() |
    df_cleaned["Description"]
        .astype("string")
        .str.strip()
        .eq("")
        .fillna(False)
)

# Create a separate analysis-friendly description
# while preserving the original column
df_cleaned["Description_Clean"] = (
    df_cleaned["Description"]
    .astype("string")
    .str.strip()
    .fillna("Unknown Product")
)

df_cleaned["Description_Clean"] = (
    df_cleaned["Description_Clean"]
    .replace("", "Unknown Product")
)

# Flag missing/invalid invoice dates
df_cleaned["Invalid_InvoiceDate"] = (
    df_cleaned["InvoiceDate"].isna()
)

print("FLOWFORGE - MISSING VALUE TREATMENT")

print("\nMissing CustomerIDs:",
      df_cleaned["Missing_CustomerID"].sum())

print("Missing Descriptions:",
      df_cleaned["Missing_Description"].sum())

print("Invalid Invoice Dates:",
      df_cleaned["Invalid_InvoiceDate"].sum())

print("\nDataset Preview:")
display(
    df_cleaned[
        [
            "InvoiceNo",
            "CustomerID",
            "Description",
            "Description_Clean",
            "Missing_CustomerID",
            "Missing_Description",
            "Invalid_InvoiceDate"
        ]
    ].head(10)
)
```

```python

# ====================================================
# FLOWFORGE - STEP 5.3
# DUPLICATE HANDLING
# ====================================================

import pandas as pd

# Original columns used for duplicate detection
original_columns = raw_df.columns.tolist()

# Identify exact duplicates based on original fields
df_cleaned["Is_Duplicate"] = (
    df_cleaned.duplicated(
        subset=original_columns,
        keep="first"
    )
)

# Count duplicate records
duplicate_count = df_cleaned["Is_Duplicate"].sum()

duplicate_percentage = (
    duplicate_count / len(df_cleaned) * 100
)

print("=" * 50)
print("FLOWFORGE - DUPLICATE HANDLING")
print("=" * 50)

print("Total Records:", len(df_cleaned))
print("Duplicate Records:", duplicate_count)
print(
    "Duplicate Percentage:",
    round(duplicate_percentage, 2), "%"
)

# Preview duplicate records
print("\nSample Duplicate Records:")
display(
    df_cleaned[
        df_cleaned["Is_Duplicate"]
    ].head(10)
)
```

```python

# ====================================================
# FLOWFORGE - FIX: CREATE df_final
# ====================================================

import pandas as pd

# Check that the preprocessed dataset exists
if "df_cleaned" not in globals():
    print("ERROR: df_cleaned is missing.")
    print("Please rerun Steps 5.1 and 5.2.")
else:
    # Use original transaction columns for duplicate checks
    original_columns = [
        "InvoiceNo", "StockCode", "Description",
        "Quantity", "InvoiceDate", "UnitPrice",
        "CustomerID", "Country"
    ]

    # Flag exact duplicate records
    df_cleaned["Is_Duplicate"] = df_cleaned.duplicated(
        subset=original_columns,
        keep="first"
    )

    # Preserve duplicates separately
    df_duplicates = df_cleaned[
        df_cleaned["Is_Duplicate"]
    ].copy()

    # Create deduplicated dataset
    df_final = df_cleaned[
        ~df_cleaned["Is_Duplicate"]
    ].copy()

    df_final.reset_index(drop=True, inplace=True)

    print("df_final created successfully!")
    print("Original records:", len(df_cleaned))
    print("Duplicate records:", len(df_duplicates))
    print("Remaining records:", len(df_final))

    display(df_final.head())

# ====================================================
# FLOWFORGE - DUPLICATE CLEANING VISUALIZATION
# ====================================================

import matplotlib.pyplot as plt

labels = [
    "Before Cleaning",
    "After Cleaning"
]

counts = [
    len(df_cleaned),
    len(df_final)
]

plt.figure(figsize=(8, 5))

bars = plt.bar(
    labels,
    counts,
    color=["#E67E22", "#27AE60"]
)

plt.title("FlowForge - Duplicate Removal Comparison")
plt.ylabel("Number of Records")

for bar in bars:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        height,
        f"{int(height):,}",
        ha="center",
        va="bottom"
    )

plt.ylim(0, max(counts) * 1.10)
plt.tight_layout()
plt.show()
```

```python

# ====================================================
# FLOWFORGE - STEP 5.4
# FEATURE ENGINEERING
# ====================================================

import pandas as pd

# Ensure invoice date is datetime
df_final["InvoiceDate"] = pd.to_datetime(
    df_final["InvoiceDate"],
    errors="coerce"
)

# Calculate transaction line amount
df_final["TotalAmount"] = (
    df_final["Quantity"] * df_final["UnitPrice"]
)

# Extract date features
df_final["Invoice_Year"] = df_final["InvoiceDate"].dt.year
df_final["Invoice_Month"] = df_final["InvoiceDate"].dt.month
df_final["Invoice_Day"] = df_final["InvoiceDate"].dt.day
df_final["Invoice_Hour"] = df_final["InvoiceDate"].dt.hour

df_final["Day_of_Week"] = (
    df_final["InvoiceDate"].dt.day_name()
)

print("FLOWFORGE - FEATURE ENGINEERING COMPLETED")

display(
    df_final[
        [
            "InvoiceNo",
            "InvoiceDate",
            "Quantity",
            "UnitPrice",
            "TotalAmount",
            "Invoice_Year",
            "Invoice_Month",
            "Invoice_Day",
            "Invoice_Hour",
            "Day_of_Week"
        ]
    ].head(10)
)
```
